```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import KFold, cross_val_score, cross_val_predict

from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
from google.colab import files

uploaded = files.upload()

df = pd.read_csv('donnees_regression_lineaire_multiple.csv')

print(df.head())
```

```python
X = df[['note math', 'note physique', 'note français']]

y = df['note finale']
```

```python
kf = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
model = LinearRegression()

scores = cross_val_score(
    model,
    X,
    y,
    cv=kf,
    scoring='r2'
)

print("R2 Score pour chaque Fold :")

for i, score in enumerate(scores):
    print(f"Fold {i+1} : {score:.4f}")

print("Moyenne R2 :", scores.mean())
```

```python
y_pred = cross_val_predict(model, X, y, cv=kf)

print("Les prédictions :")
print(y_pred)
```

```python
mse = mean_squared_error(y, y_pred)

rmse = np.sqrt(mse)

mae = mean_absolute_error(y, y_pred)

r2 = r2_score(y, y_pred)

print("MSE :", mse)
print("RMSE :", rmse)
print("MAE :", mae)
print("R2 Score :", r2)
```

```python
import matplotlib.pyplot as plt

plt.scatter(y, y_pred)

plt.xlabel("Valeurs réelles")
plt.ylabel("Valeurs prédites")

plt.title("Régression Linéaire Multiple avec K-Fold")

plt.show()
```

```python
import pandas as pd
from sklearn.model_selection import KFold
from sklearn.linear_model import LinearRegression
import numpy as np

df = pd.read_csv('donnees_regression_lineaire_multiple.csv')
X = df[['note math', 'note physique', 'note français']]
y = df['note finale']

kf = KFold(n_splits=5, shuffle=True, random_state=42)
model = LinearRegression()

y_pred_all = np.zeros_like(y)

for train_index, test_index in kf.split(X):
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]

    model.fit(X_train, y_train)
    y_pred_all[test_index] = model.predict(X_test)

resultats_df = df.copy()
resultats_df['note predite'] = y_pred_all

print(resultats_df[['note math', 'note physique', 'note français', 'note finale', 'note predite']].head(15))

resultats_df.to_csv('resultats_predictions.csv', index=False)
```
